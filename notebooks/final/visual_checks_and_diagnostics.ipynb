#### Keyword analysis

```python
from src.text_processing import process_keywords
import pandas as pd

# Load data after title matching
crs_matched = pd.read_feather("../../data/processed/title_matched/crs_titles_matched_wo_stopwords.feather")

# Left join the matched keywords on the 'project_title' column
crs_merged = crs.merge(crs_matched, on='project_title', how='left', )

keywords_file_path = "../../data/keywords/statistics_gender/keyword_statistics_gender.xlsx"
stat_keywords = pd.read_excel(keywords_file_path, sheet_name="statistics")
gen_keywords = pd.read_excel(keywords_file_path, sheet_name="gender")

stat_keywords = process_keywords(stat_keywords, remove_stopwords=False)
gen_keywords = process_keywords(gen_keywords, remove_stopwords=False)
```

```python
crs_for_keywords = crs_merged[["year", "project_title", "language", "stat_keywords", "gen_keywords"]]

crs_for_keywords = crs_for_keywords[crs_for_keywords["year"] == 2023]
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Count matched keywords by concept ID and language
# Build a lookup table from keyword text to concept ID, language, and English label
id_column = next((col for col in stat_keywords.columns if col.lower() in {"id", "keyword_id"}), None)
if id_column is None:
    raise ValueError("stat_keywords must contain an ID column named 'id' or 'keyword_id'.")

language_columns = [col for col in ["en", "fr", "es", "de"] if col in stat_keywords.columns]
keyword_lookup = []

for _, row in stat_keywords.iterrows():
    keyword_id = row[id_column]
    if pd.isna(keyword_id):
        continue

    english_keyword = row.get("en")
    if pd.isna(english_keyword):
        english_keyword = next((row[lang] for lang in language_columns if pd.notna(row[lang])), None)

    for language in language_columns:
        keyword = row[language]
        if pd.notna(keyword):
            keyword_lookup.append({
                "keyword": keyword,
                "keyword_id": keyword_id,
                "language": language,
                "english_keyword": english_keyword,
            })

keyword_lookup = pd.DataFrame(keyword_lookup).drop_duplicates(subset=["keyword"], keep="first")

# Map matched keywords back to their concept IDs and languages
matched_keyword_counts = (
    crs_for_keywords["stat_keywords"]
    .explode()
    .dropna()
    .rename("keyword")
    .to_frame()
    .merge(keyword_lookup, on="keyword", how="left")
    .dropna(subset=["keyword_id"])
    .groupby(["keyword_id", "english_keyword", "language"])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=language_columns, fill_value=0)
    .sort_index()
 )

# Order by total matched keywords, highest count first so it appears at the lowest y value
matched_keyword_totals = matched_keyword_counts.sum(axis=1).sort_values(ascending=False)
matched_keyword_counts = matched_keyword_counts.loc[matched_keyword_totals.index]

# Assign colors to each language
language_colors = {"en": "steelblue", "fr": "tomato", "es": "mediumseagreen", "de": "goldenrod"}

# Plot stacked horizontal bars: one row per keyword concept, colored by language counts
fig, ax = plt.subplots(figsize=(12, max(7, 0.35 * len(matched_keyword_counts))))
bar_height = 0.55
row_step = 0.72
left = np.zeros(len(matched_keyword_counts))
y_positions = np.arange(len(matched_keyword_counts)) * row_step

for language in language_columns:
    values = matched_keyword_counts[language].to_numpy()
    ax.barh(
        y_positions,
        values,
        left=left,
        height=bar_height,
        color=language_colors.get(language, "gray"),
        label=language.upper(),
    )
    left += values

# Annotate total counts at the end of each stacked bar
for y_pos, total in zip(y_positions, matched_keyword_totals.loc[matched_keyword_counts.index]):
    ax.text(total + 0.1, y_pos, str(int(total)), va="center", fontsize=9)

ax.set_yticks(y_positions)
ax.set_yticklabels(matched_keyword_counts.index.get_level_values("english_keyword").astype(str), fontsize=9)
ax.set_xlabel("Count")
ax.set_ylabel("Keyword")
ax.set_title(f"Number of matched Keywords (Total matched keywords: {int(matched_keyword_totals.sum())})")
ax.set_ylim(-bar_height / 2 - 0.1, y_positions[-1] + bar_height / 2)
ax.margins(y=0)
ax.legend(title="Language", loc="upper right")
plt.tight_layout()
plt.show()
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt

def extract_keywords(value):
    # Handles real lists/arrays and string-encoded arrays like "['k1' 'k2']"
    if isinstance(value, list):
        return [k for k in value if isinstance(k, str) and k.strip()]
    if isinstance(value, np.ndarray):
        return [str(k).strip() for k in value.tolist() if str(k).strip()]
    if isinstance(value, str):
        stripped = value.strip()
        if not stripped or stripped in {"[]", "[ ]"}:
            return []
        # Extract quoted tokens from formats such as ['a' 'b'] or ["a", "b"]
        matches = re.findall(r"'([^']+)'|\"([^\"]+)\"", stripped)
        tokens = [m[0] or m[1] for m in matches]
        if tokens:
            return [t.strip() for t in tokens if t.strip()]
        # Fallback: split loose text
        stripped = stripped.strip("[]")
        return [t.strip().strip("'\"") for t in re.split(r"[,\s]+", stripped) if t.strip()]
    return []

# Parse keywords into a normalized list column
parsed_keywords = crs_for_keywords["stat_keywords"].apply(extract_keywords)

# Keep only rows where exactly one statistics keyword was detected
single_keyword_projects = crs_for_keywords[parsed_keywords.map(len) == 1].copy()
single_keyword_projects["keyword"] = parsed_keywords[parsed_keywords.map(len) == 1].str[0]

# Count how many project rows were detected by each concept ID and keyword language
project_keyword_counts = (
    single_keyword_projects
    .merge(
        keyword_lookup.rename(columns={"language": "keyword_language"}),
        on="keyword",
        how="left",
    )
    .dropna(subset=["keyword_id"])
    .groupby(["keyword_id", "english_keyword", "keyword_language"])
    .size()
    .unstack(fill_value=0)
    .reindex(columns=language_columns, fill_value=0)
    .sort_index()
 )

# Order by total detected projects, highest count first so it appears at the lowest y value
project_keyword_totals = project_keyword_counts.sum(axis=1).sort_values(ascending=False)
project_keyword_counts = project_keyword_counts.loc[project_keyword_totals.index]

# Plot stacked horizontal bars: one row per keyword concept, colored by project counts per language
fig, ax = plt.subplots(figsize=(12, max(7, 0.35 * len(project_keyword_counts))))
bar_height = 0.55
row_step = 0.72
left = np.zeros(len(project_keyword_counts))
y_positions = np.arange(len(project_keyword_counts)) * row_step

for language in language_columns:
    values = project_keyword_counts[language].to_numpy()
    ax.barh(
        y_positions,
        values,
        left=left,
        height=bar_height,
        color=language_colors.get(language, "gray"),
        label=language.upper(),
    )
    left += values

# Annotate total project counts at the end of each stacked bar
for y_pos, total_projects in zip(y_positions, project_keyword_totals.loc[project_keyword_counts.index]):
    ax.text(total_projects + 0.1, y_pos, str(int(total_projects)), va="center", fontsize=9)

ax.set_yticks(y_positions)
ax.set_yticklabels(project_keyword_counts.index.get_level_values("english_keyword").astype(str), fontsize=9)
ax.set_xlabel("Count")
ax.set_ylabel("Keyword")
ax.set_title(f"Number of Projects detected fory every keyword (only projects with one keyword detected -> contribution from specific keyword)")
ax.set_ylim(-bar_height / 2, y_positions[-1] + bar_height / 2)
ax.margins(y=0)
ax.legend(title="Language", loc="upper right")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
thresholds = [round(x, 7) for x in np.arange(0.999709, 0.999408, -0.00001)]

thresholds
```

```python
# Aggregate the the total of usd_disbursement grouped by year and make one line for each threshold 
def aggregate_by_threshold(df, thresholds):
    results = []
    for threshold in thresholds:
        # Filter the DataFrame based on the threshold
        filtered_df = df[df['probability_is_statistics'] >= threshold]
        
        # Group by year and sum the usd_disbursement 
        aggregated = filtered_df.groupby('year')['usd_disbursement'].sum().reset_index()
        
        # Add the threshold to the aggregated DataFrame
        aggregated['threshold'] = threshold
        
        results.append(aggregated)
    
    return pd.concat(results, ignore_index=True)

# Apply the aggregation function to the CRS DataFrame
aggregated_results = aggregate_by_threshold(crs, thresholds)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

plt.figure(figsize=(14, 8))

# Use a qualitative colormap for distinct colors
cmap = plt.get_cmap('tab20', len(thresholds))
colors = [cmap(i) for i in range(len(thresholds))]

# To avoid overlap, stagger the annotation positions
offsets = np.linspace(-50, 50, len(thresholds))

for idx, threshold in enumerate(thresholds):
    threshold_data = aggregated_results[aggregated_results['threshold'] == threshold]
    plt.plot(threshold_data['year'], threshold_data['usd_disbursement'], label=f'{threshold}', color=colors[idx])
    # Annotate the last year value to the right of the line
    if not threshold_data.empty:
        last_row = threshold_data.iloc[-1]
        plt.text(
            last_row['year'] + 0.2,  # Slightly to the right
            last_row['usd_disbursement'] + offsets[idx],  # Stagger vertically
            f"{int(last_row['usd_disbursement'])}M",  # Format as millions
            va='center',
            fontsize=8,
            color=colors[idx]
        )

plt.title('USD Disbursement by Year for Different Thresholds')
plt.xlabel('Year')
plt.ylabel('USD Disbursement (in millions)')
plt.xticks(threshold_data['year'].unique())
plt.legend(title='Thresholds', bbox_to_anchor=(1.05, 1), loc='upper left', ncol=1)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

year = 2024

# Keep relevant columns and clean missing values
donor_data = press.loc[press["year"] == year, ["donor_name", "donor_type"]].copy()
donor_data["donor_name"] = donor_data["donor_name"].fillna("Unknown donor")
donor_data["donor_type"] = donor_data["donor_type"].fillna("Unknown donor type")

# Count projects per donor and sort descending so the largest donor is at low x values
projects_by_donor = donor_data["donor_name"].value_counts().sort_values(ascending=False)

# Assign one donor_type per donor (mode in case of inconsistent labels)
donor_type_by_donor = donor_data.groupby("donor_name")["donor_type"].agg(
    lambda x: x.mode().iloc[0] if not x.mode().empty else "Unknown donor type"
)
ordered_donor_types = donor_type_by_donor.reindex(projects_by_donor.index).fillna("Unknown donor type")

# Build a softer, visually balanced color map by donor_type
unique_donor_types = sorted(ordered_donor_types.unique().tolist())
nice_palette = [
    "#4E79A7",  # muted blue
    "#59A14F",  # muted green
    "#E15759",  # muted red
    "#F28E2B",  # soft orange
    "#76B7B2",  # teal
    "#B07AA1",  # mauve
    "#EDC948",  # mustard
    "#9C755F",  # brown
    "#BAB0AC",  # gray
    "#86BCB6",  # light teal
]
donor_type_colors = {
    dtype: nice_palette[i % len(nice_palette)] for i, dtype in enumerate(unique_donor_types)
}
bar_colors = ordered_donor_types.map(donor_type_colors)

plt.figure(figsize=(max(14, 0.2 * len(projects_by_donor)), 7))
bars = plt.bar(
    projects_by_donor.index,
    projects_by_donor.values,
    color=bar_colors,
    edgecolor="black",
    width=0.4,
)

# Add totals above each donor bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height)}",
        ha="center",
        va="bottom",
        fontsize=8,
    )

# Add legend for donor_type colors
legend_handles = [
    Patch(facecolor=donor_type_colors[dtype], edgecolor="black", label=dtype)
    for dtype in unique_donor_types
]
plt.legend(handles=legend_handles, title="donor_type", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.title("Number of Projects by Donor for year {} (total projects: {})".format(year, projects_by_donor.sum()))
plt.xlabel("Donor")
plt.ylabel("Number of Projects")
plt.xticks(rotation=45, fontsize=8, ha="right")
plt.margins(x=0.01)
plt.tight_layout()
plt.show()
```
